-The dates are written in different ways.
-Some prices are empty.
-Some zip codes are missing the first 0.

```python
df.isna().sum()
```

```python
fill_value = df["price"].median()
print(fill_value)
df["price"] = df["price"].fillna(fill_value)
df.isna().sum()
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
df.shape
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
df["zip"].head(10)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
df.dtypes
```

```python
df[df["qty"] < 0]
```

```python
df["qty"] = df["qty"].abs()
print(df.loc[[202, 262, 297]])
df.shape
```

```python
df.to_csv("sales_clean.csv", index=False)
```

**Cleaning Log**
- I started with 300 rows of sales data.
- 12 prices were empty. I filled them with the median price, 37.53, because the expensive prices change the mean.
- I found 8 rows that were exactly the same and deleted them. Now there are 292 rows.
- Some zip codes lost their first 0, so I added it back because the addresses were wrong without 0.
- The dates were written in different styles, so I made them all the same date type.
- 3 rows had a minus quantity. I treated them as typos and changed them to positive because minuses are impossible.
- If this cleaning is done wrong, the company might make a wrong decision.
