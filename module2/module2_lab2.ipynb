```python
def multiply_for(a, b):
    result = 0
    for i in range(b):
        result = result + a
    return result
```

```python
def multiply_while(a, b):
    result = 0
    count = 0
    while count < b:
        result = result + a
        count = count + 1
    return result
```

```python
tests = [(7, 6), (3, 4), (5, 1), (10, 0)]
for a, b in tests:
    print(f"{a} x {b}: for = {multiply_for(a, b)}, while = {multiply_while(a, b)}, real = {a * b}")
```

The while loop stops because count goes up by 1 each time. When count reaches b, the condition count < b is false, so the loop ends.
