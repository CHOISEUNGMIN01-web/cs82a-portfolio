```python
import pandas as pd
df = pd.read_csv("sales_clean.csv")
df.shape
```

```python
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())
```

```python
print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```

The mean is higher because a few expensive items pull it up. So I think the median is a better way to show the typical price.

```python
ax = df["price"].plot(kind="hist", bins=20, title="Price Distribution")
ax.set_xlabel("Price")
ax.set_ylabel("Number of orders")
```

The histogram of price is right-skewed because it decreases when it goes to the right

```python
df["revenue"] = df["price"] * df["qty"]
df.head()
```

```python
df.corr(numeric_only=True)
```

```python
ax = df.plot.scatter(x="price", y="revenue", title="Price vs Revenue")
ax.set_xlabel("Price")
ax.set_ylabel("Revenue")
```

ax = df.plot.scatter(x="qty", y="revenue", title="Quantity vs Revenue")
ax.set_xlabel("Quantity")
ax.set_ylabel("Revenue")

Orders with higher prices usually make more revenue. The link is strong (r = 0.82). Quantity has only a weak link to revenue (r = 0.31). But a few very expensive orders (over $250) may make the link look stronger.
